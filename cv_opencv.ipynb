```python
import cv2
```

```python
from google.colab.patches import cv2_imshow
image = cv2.imread("/content/cat_image.jpeg")
cv2_imshow( image)
```

```python
print("SHAPE:", image.shape)
#739 height
#415 width
#color RGB urniga BGR GREEN BLUE
print("HEIGTH", image.shape[0])
print("WIDTH", image.shape[1])
print("CHANELS", image.shape[2])
```

```python
(b, g, r)= image[50, 100]
print("blue:", b, "red:", g, "red:", r )
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2_imshow(gray)
```

```python
resized = cv2.resize(gray,(300,300))
cv2_imshow(resized)
```

```python
cropped = image[200:300, 150:300]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(gray, 200)
cv2_imshow(flipped)
```

```python
cv2.imwrite("flipped_cat.jpeg", flipped)
```

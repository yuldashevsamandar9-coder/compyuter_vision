```python
from bing_image_downloader import downloader
from google.colab import drive
import os, shutil

drive.mount('/content/drive')
```

```python
downloader.download(
    "Uzbek plov",
    limit=100,
    output_dir="/content/pilaf_images",
    adult_filter_off=False,
    force_replace=False,
    timeout=60
)
```

```python
source_dir = '/content/pilaf_images/Uzbek plov'
target_dir = '/content/drive/MyDrive/food101_dataset/pilaf'

os.makedirs(target_dir, exist_ok=True)

for filename in os.listdir(source_dir):
    src_file = os.path.join(source_dir, filename)
    dst_file = os.path.join(target_dir, filename)
    shutil.copy2(src_file, dst_file)
```

```python
downloader.download(
    "kebab shashlik food",
    limit=100,
    output_dir="/content/kebab_images",
    adult_filter_off=False,
    force_replace=False,
    timeout=60
)
```

```python
source_dir = '/content/kebab_images/kebab shashlik food'
target_dir = '/content/drive/MyDrive/food101_dataset/kebab'

os.makedirs(target_dir, exist_ok=True)

for filename in os.listdir(source_dir):
    src_file = os.path.join(source_dir, filename)
    dst_file = os.path.join(target_dir, filename)
    shutil.copy2(src_file, dst_file)
```

```python
# Drive dagi rasmlar soni
for food in ['pilaf', 'kebab']:
    folder = '/content/drive/MyDrive/food101_dataset/' + food
    images = [name for name in os.listdir(folder) if name.endswith('.jpg')]
    print(food, ':', len(images), 'ta rasm')
```
